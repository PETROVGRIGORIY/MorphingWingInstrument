```python
import dafeaero.xfoilpy as da
```

```python
airfoil1 = da.Airfoil("SG6041", "dafeaero/airfoils/foildatabase/SG6041.dat")
airfoil2 = da.Airfoil("SG6042", "dafeaero/airfoils/foildatabase/SG6042.dat")
airfoil3 = da.Airfoil("SG6043", "dafeaero/airfoils/foildatabase/SG6043.dat")

foil_comparator = da.AirfoilComparator()
foil_comparator.add_airfoil(airfoil1)
foil_comparator.add_airfoil(airfoil2)
foil_comparator.add_airfoil(airfoil3)
```

## Режим: сравнение на одном графике

```python
foil_comparator.cl_alpha_plot(
    300000, 0.1, 9, 500, da.AnglesOfAttackSequence(-6, 20, 0.5),
    ("SG6041", "SG6042", "SG6043"), da.ComparisonMode.COMBINED,
    "Исследование профилей SG")
```

## Режим: сравнение на разных графиках

```python
foil_comparator.cl_alpha_plot(
    300000, 0.1, 9, 500, da.AnglesOfAttackSequence(-6, 20, 0.5),
    ("SG6041", "SG6042", "SG6043"), da.ComparisonMode.SEPARATE, "Исследование профилей SG")
```
