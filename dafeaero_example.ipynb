## Классы Airfoil и Polar

```python
airfoil = da.Airfoil.naca("2612")

print(f"Кривизна профиля: {airfoil.camber:.4f}")
print(f"Положение максимальной кривизны: {airfoil.camber_position:.4f}")
print(f"Толщина профиля: {airfoil.thickness:.4f}")
print(f"Положение максимальной толщины: {airfoil.thickness_position * 100:.1f}% хорды")
```

```python
airfoil.plot()

with_flap = airfoil.with_flap(20, 0.7)
with_flap.plot()
```

```python
polar = airfoil.polar(300000, 0.1, da.AnglesOfAttackSequence(0, 25, 1))
print(f"максимально качество: {polar.ld_max}")
print(f"угол атаки максимального качества: {polar.ld_max_alpha}")
```

## Класс AirfoilComparator

```python
airfoil1 = da.Airfoil.naca(2412)
airfoil2 = da.Airfoil.naca(2612)
airfoil3 = da.Airfoil.naca(2812)
 
foil_comparator = da.AirfoilComparator()
foil_comparator.add_airfoil(airfoil1)
foil_comparator.add_airfoil(airfoil2)
foil_comparator.add_airfoil(airfoil3)
```

### Режим: сравнение на одном графике

```python
foil_comparator.cl_alpha_plot(
    300000, 0.1, 9, 500, da.AnglesOfAttackSequence(-6, 20, 0.5),
    "", False,
    "Исследование профилей SG")
```

### Режим: сравнение на разных графиках

```python
foil_comparator.cd_alpha_plot(
    300000, 0.1, 9, 500, da.AnglesOfAttackSequence(-6, 20, 0.5),
    "", True, "Исследование профилей SG")
```
